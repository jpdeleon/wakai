## Fbol, Luminosity, Teff, Rstar

Method
* derive bolometric flux, F_bol, from the integral of the absolutely calibrated spectrum, 
* derive luminosity, L_*, from F_bol and the Gaia distance,
* derive Teff from comparing the calibrated spectrum to atmospheric models
* derive stellar radius R* from the ratio of the absolutely calibrated spectrum to the model (equal to R2/D2; Blackwell & Shallis 1977).

Data
* optical and near-IR (NIR) photometry from the Two Micron All Sky Survey (2MASS; Skrutskie et al. 2006),
* Wide-field Infrared Survey Explorer (WISE; Cutri et al. 2013),
* Gaia data release 2 (DR2; Evans et al. 2018; Gaia Collaboration et al. 2018), and
* Tycho-2 (Høg et al. 2000).

Errors
* to account for the variability of the source, we assume that all photometry has an addition error of 0.02 mag (for optical) or 0.01 mag (for NIR)

Validity

* For FGK stars, use spectral template from (Baraffe et al. 2015) and photometry from Gaia DR2, 2MASS, WISE, and Tycho-2 
* For M dwarf stars, use optical and NIR spectral templates from Gaidos et al. (2014) and photometry from Gaia DR2, 2MASS, and WISE.
