## Vizier

```python
from astroquery.vizier import Vizier
from astropy.coordinates import SkyCoord
import astropy.units as u

# Setup Vizier to query Gaia DR3
v = Vizier(columns=['*'], row_limit=-1)
# 'I/355/gaiadr3' is the Vizier ID for Gaia DR3
result = v.query_region(target_coord, 
                        radius=query_radius, 
                        catalog='I/355/gaiadr3')

print(result[0])
```
