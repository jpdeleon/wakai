## cross-match

### young planets

```python
df_cp = pd.read_csv('/ut2/jerome/github/research/project/TOI-6715/data/young_validated_planets_TOIs - validated_hosts.csv',
                    skiprows=[1],
                    dtype={'RA': float, 'DEC': float}
                   )
df_cp = df_cp.dropna(subset=['RA','DEC'])
```

```python
df_cp.columns
```

```python
sc1 = SkyCoord(ra=df_cp['RA'].values*u.deg, dec=df_cp['DEC'].values*u.deg)
sc2 = SkyCoord(ra=df2['RAdeg'].values*u.deg, dec=df2['DEdeg'].values*u.deg)
idx, d2d, d3d = sc1.match_to_catalog_sky(sc2)
```

```python
max_sep = 1 * u.arcsec
mask = d2d < max_sep

# Keep only the rows from df1 that have a match within the threshold
df1_matches = df_cp[mask].copy()

# Add the matched data from df2 to these rows
# We use the 'idx' array filtered by the same mask
df2_matches = df2.iloc[idx[mask]].reset_index(drop=True)

# Combine them into a single crossmatched dataframe
df_final = pd.concat([df1_matches.reset_index(drop=True), df2_matches], axis=1)
```

### clusters in Perren+2023

```python
catalog_name = 'Perren2023'
df_cat = pd.read_csv('../data/Perren2023/UCC_cat.csv')
df_cat.tail()
```

```python
df_mem = pd.read_parquet('../data/Perren2023/UCC_members.parquet').dropna()
df_mem.tail()
```

```python
sc1 = SkyCoord(ra=df_mem['RA_ICRS'].values*u.deg, dec=df_mem['DE_ICRS'].values*u.deg)
sc2 = SkyCoord(ra=df2['RAdeg'].values*u.deg, dec=df2['DEdeg'].values*u.deg)
```

```python
idx, d2d, d3d = sc1.match_to_catalog_sky(sc2)
```

```python
max_sep = 3 * u.arcsec
mask = d2d < max_sep

# Keep only the rows from df1 that have a match within the threshold
df1_matches = df_mem[mask].copy()

# Add the matched data from df2 to these rows
# We use the 'idx' array filtered by the same mask
df2_matches = df2.iloc[idx[mask]].reset_index(drop=True)

# Combine them into a single crossmatched dataframe
df_final = pd.concat([df1_matches.reset_index(drop=True), df2_matches], axis=1)
```
